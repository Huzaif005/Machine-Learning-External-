```python
df = pd.read_csv("Position_Salaries.csv")

print(df)
```

```python
X = df.iloc[:, 0:1].values  # Select first column
y = df.iloc[:, 1].values    # Select second column

print("X values")
print(X)

print("Y values")
print(y)
```

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

X = df.iloc[:, 1:2].values
y = df.iloc[:, 2].values

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

model = LinearRegression()
model.fit(X_poly, y)

y_pred = model.predict(X_poly)

print("Polynomial Regression R2 Score =", r2_score(y, y_pred))
```

```python
poly2 = PolynomialFeatures(degree=2)

X_poly2 = poly2.fit_transform(X)

model2 = LinearRegression()

model2.fit(X_poly2,y)

pred2 = model2.predict(X_poly2)

print("Degree 2 R2 Score =",r2_score(y,pred2))
```

```python
poly3 = PolynomialFeatures(degree=3)

X_poly3 = poly3.fit_transform(X)

model3 = LinearRegression()

model3.fit(X_poly3,y)

pred3 = model3.predict(X_poly3)

print("Degree 3 R2 Score =",r2_score(y,pred3))
```

```python
poly4 = PolynomialFeatures(degree=4)

X_poly4 = poly4.fit_transform(X)

model4 = LinearRegression()

model4.fit(X_poly4,y)

pred4 = model4.predict(X_poly4)

print("Degree 4 R2 Score =",r2_score(y,pred4))
```

```python
import numpy as np
import matplotlib.pyplot as plt

X = df.iloc[:, 1:2].values  # Level
y = df.iloc[:, 2].values    # Salary

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

poly4 = PolynomialFeatures(degree=4)
X_poly4 = poly4.fit_transform(X)

model4 = LinearRegression()
model4.fit(X_poly4, y)

X_grid = np.arange(X.min(), X.max(), 0.1)
X_grid = X_grid.reshape(-1, 1)

plt.scatter(X, y, color="red", label="Actual Salary")

plt.plot(
    X_grid,
    model4.predict(poly4.transform(X_grid)),
    label="Degree 4 Polynomial"
)

plt.title("Polynomial Regression Degree 4")
plt.xlabel("Level")
plt.ylabel("Salary")
plt.legend()
plt.show()
```
